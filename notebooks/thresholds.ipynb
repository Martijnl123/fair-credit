```python
import sys
sys.path.insert(0, "../src")

import numpy as np
from fair_credit.data import load
from fair_credit.pipeline import split, make_model

ds = load("taiwan")
tr, va, te = split(ds, seed=0)
X = ds.X.to_numpy()
clf = make_model("lr", seed=0).fit(X[tr], ds.y[tr])
p_va = clf.predict_proba(X[va])[:, 1]
```

```python
from sklearn.metrics import roc_curve

def choose_threshold(y, p):
    fpr, tpr, thresholds = roc_curve(y, p)
    best = np.argmax(tpr - fpr)
    return float(thresholds[best])
```

```python
t = choose_threshold(ds.y[va], p_va)
print("threshold:", round(t, 3))
print("approval at 0.5:", round((p_va >= 0.5).mean(), 3))
print("approval at t:  ", round((p_va >= t).mean(), 3))
```
